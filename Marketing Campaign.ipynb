```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
print("Average ROI:", df["ROI"].mean())
```

```python
print("Minimum ROI:", df["ROI"].min())
```

```python

print("Average Conversion Rate:", df["Conversion_Rate"].mean())
```

```python
channel = df["Channel_Used"].value_counts()

plt.figure(figsize=(8,5))
channel.plot(kind="bar")
plt.title("Campaigns by Channel")
plt.xlabel("Channel")
plt.ylabel("Count")
plt.show()
```

```python
roi = df.groupby("Channel_Used")["ROI"].mean()

plt.figure(figsize=(8,5))
roi.plot(kind="bar")
plt.title("Average ROI by Channel")
plt.xlabel("Channel")
plt.ylabel("ROI")
plt.show()
```

```python
conversion = df.groupby("Channel_Used")["Conversion_Rate"].mean()

plt.figure(figsize=(8,5))
conversion.plot(kind="bar")
plt.title("Average Conversion Rate by Channel")
plt.xlabel("Channel")
plt.ylabel("Conversion Rate")
plt.show()
```

```python
cost = df.groupby("Channel_Used")["Acquisition_Cost"].mean()

plt.figure(figsize=(8,5))
cost.plot(kind="bar")
plt.title("Average Acquisition Cost")
plt.xlabel("Channel")
plt.ylabel("Cost")
plt.show()
```

```python
goal = df["Campaign_Goal"].value_counts()

plt.figure(figsize=(7,7))
goal.plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Campaign Goals")
plt.show()
```

```python
segment = df["Customer_Segment"].value_counts()

plt.figure(figsize=(8,5))
segment.plot(kind="bar")
plt.title("Customer Segment")
plt.show()
```

```python
audience = df["Target_Audience"].value_counts()

plt.figure(figsize=(10,5))
audience.plot(kind="bar")
plt.title("Target Audience")
plt.show()
```

```python
company = df["Company"].value_counts()

plt.figure(figsize=(10,5))
company.plot(kind="bar")
plt.title("Company Campaign Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["ROI"], bins=20)
plt.title("ROI Distribution")
plt.xlabel("ROI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Clicks"], bins=20)
plt.title("Clicks Distribution")
plt.xlabel("Clicks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Impressions"], bins=20)
plt.title("Impressions Distribution")
plt.xlabel("Impressions")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Engagement_Score"], bins=20)
plt.title("Engagement Score")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
plt.figure(figsize=(10,8))

plt.imshow(corr, cmap="coolwarm", aspect="auto")

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)

plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Heatmap")

plt.show()
```
